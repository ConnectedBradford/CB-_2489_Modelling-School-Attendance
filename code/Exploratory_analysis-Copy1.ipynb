# Split cohorts, rename to Year7-11 and rejoin

```r
cohort1 <- raw_data |>
    filter(cohort == '1998/99') |>
    select(-c(Y2015.2016_Term1, Y2015.2016_Term2, Y2015.2016_Term3)) |>
    rename(
        Year7_Term1 = Y2010.2011_Term1,
        Year7_Term2 = Y2010.2011_Term2,
        Year7_Term3 = Y2010.2011_Term3,
        Year8_Term1 = Y2011.2012_Term1,
        Year8_Term2 = Y2011.2012_Term2,
        Year8_Term3 = Y2011.2012_Term3,
        Year9_Term1 = Y2012.2013_Term1,
        Year9_Term2 = Y2012.2013_Term2,
        Year9_Term3 = Y2012.2013_Term3,
        Year10_Term1 = Y2013.2014_Term1,
        Year10_Term2 = Y2013.2014_Term2,
        Year10_Term3 = Y2013.2014_Term3,
        Year11_Term1 = Y2014.2015_Term1,
        Year11_Term2 = Y2014.2015_Term2,
        Year11_Term3 = Y2014.2015_Term3
        )
```

```r
head(cohort1)
```

```r
cohort2 <- raw_data |>
    filter(cohort == '1999/00') |>
    select(-c(Y2010.2011_Term1, Y2010.2011_Term2, Y2010.2011_Term3)) |>
    rename(
        Year7_Term1 = Y2011.2012_Term1,
        Year7_Term2 = Y2011.2012_Term2,
        Year7_Term3 = Y2011.2012_Term3,
        Year8_Term1 = Y2012.2013_Term1,
        Year8_Term2 = Y2012.2013_Term2,
        Year8_Term3 = Y2012.2013_Term3,
        Year9_Term1 = Y2013.2014_Term1,
        Year9_Term2 = Y2013.2014_Term2,
        Year9_Term3 = Y2013.2014_Term3,
        Year10_Term1 = Y2014.2015_Term1,
        Year10_Term2 = Y2014.2015_Term2,
        Year10_Term3 = Y2014.2015_Term3,
        Year11_Term1 = Y2015.2016_Term1,
        Year11_Term2 = Y2015.2016_Term2,
        Year11_Term3 = Y2015.2016_Term3
        )
```

```r
head(cohort2)
```

```r
data <- rbind(cohort1,cohort2) 
```

```r
data
```

```r
data_drop_missing <- data |>
    filter(
        Destination != 'DROP',
        !if_all(c(
                Year7_Term1, Year7_Term2, Year7_Term3,
                Year8_Term1, Year8_Term2, Year8_Term3,
                Year9_Term1, Year9_Term2, Year9_Term3,
                Year10_Term1, Year10_Term2, Year10_Term3,
                Year11_Term1, Year11_Term2, Year11_Term3
                ), is.na),
        !is.na(KS4_Entries)
        )
```

```r
colSums(is.na(data_drop_missing))
```

#### Replace svchool NAs with 111111

```r
data_drop_missing <- data_drop_missing |>
    mutate(across(c(Year7_Term1, Year7_Term2, Year7_Term3,
                Year8_Term1, Year8_Term2, Year8_Term3,
                Year9_Term1, Year9_Term2, Year9_Term3,
                Year10_Term1, Year10_Term2, Year10_Term3,
                Year11_Term1, Year11_Term2, Year11_Term3),
                ~ replace_na(., 111111)))
```

# Simplified dataset

```r
data_simple <- data_drop_missing |>
    mutate(Highest_SEN_level = case_when (
            no_SEN ~ 'No SEN', 
            ever_SEN_support & !ever_EHC_plan ~ 'SEN Support',
            ever_EHC_plan ~ 'EHC Plan'
        ))
```

```r
data_simple <- data_simple |>
    select(-c(ever_SEN_support,first_SEN_support_year,SEN_support_need,ever_EHC_plan,first_EHC_plan_year,EHC_plan_need,no_SEN))
```
