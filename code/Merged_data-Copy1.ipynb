# Absences: total, authorised, unauthorised

```r
absence_alltime <- read.csv("data/additional_absence_alltime_rates.csv", header = TRUE)
```

```r
head(absence_alltime)
```

```r
person_absence <- person |>
    left_join(absence_alltime, by=join_by(person_id))
```

```r
head(person_absence)
```

```r
person_absence <- person_absence |>
    mutate(across(c(AuthorisedAbsenceRate, UnauthorisedAbsenceRate, TotalAbsenceRate),
                ~ replace_na(., 0)))
```

```r
person_absence |>
    nrow()
```

```r
n_distinct(person_absence$person_id)
```

# Ever CIN

```r
cin <- read.csv("data/additional_ever_cin.csv", header = TRUE)
```

```r
head(cin)
```

```r
person_absence_cin <- person_absence |>
    left_join(cin, by=join_by(person_id))
```

```r
person_absence_cin <- person_absence_cin |>
    mutate(ever_CIN = case_when(
            ever_CIN ~ TRUE,
        TRUE ~ FALSE)
           )
```

```r
head(person_absence_cin)
```

```r
person_absence_cin |>
    nrow()
```

```r
n_distinct(person_absence_cin$person_id)
```

# KS4 attainment

```r
ks4 <- read.csv("data/additional_ks4_filtered.csv", header = TRUE)
```

```r
head(ks4)
```

```r
ks4 <- distinct(ks4)
```

```r
ks4 |>
    nrow()
```

```r
n_distinct(ks4$person_id)
```

```r
ks4 |>
    group_by(person_id) |>
    filter(n() > 1) |>
    arrange(person_id)
```

```r
ks4 <- ks4 |>
  arrange(person_id, KS4_IDACI) |> 
  group_by(person_id) |>
  slice_tail(n = 1) |> 
  ungroup()
```

```r
person_absence_cin_excl_ks4 <- person_absence_cin_excl |>
    left_join(ks4, by=join_by(person_id))
```

```r
head(person_absence_cin_excl_ks4)
```

```r
person_absence_cin_excl_ks4 |>
    nrow()
```

```r
n_distinct(person_absence_cin_excl_ks4$person_id)
```

```r
IDACI <- person_absence_cin_excl_ks4 |>
    select(person_id, IDACIScore_Home, KS4_IDACI)
```

```r
IDACI <- IDACI |>
    mutate(IDACI = coalesce(IDACIScore_Home, KS4_IDACI)) |>
    select(person_id, IDACI)
```

```r
person_absence_cin_excl_ks4 <- person_absence_cin_excl_ks4 |>
    left_join(IDACI, by=join_by(person_id)) |>
    select(-c(IDACIScore_Home, KS4_IDACI))
```

```r
person_absence_cin_excl_ks4 %>%
  count(person_id) %>%
  filter(n > 1)
```

```r
person_absence_cin_excl_ks4 |>
    nrow()
```

```r
n_distinct(person_absence_cin_excl_ks4$person_id)
```

# Clean and re-order cols

```r
glimpse(person_absence_cin_excl_ks4_dest)
```

```r
person_absence_cin_excl_ks4_dest_clean <- person_absence_cin_excl_ks4_dest |>
    select(cohort, 
           person_id, birth_date, gender, CombinedEthnicity, 
           EAL, ever_fsm, ever_CIN, 
           DistCurrSch, LSOA11_Home, IDACI, 
           ever_SEN_support, first_SEN_support_year, SEN_support_need, 
           ever_EHC_plan, first_EHC_plan_year, EHC_plan_need, 
           no_SEN, 
           AuthorisedAbsenceRate, UnauthorisedAbsenceRate, TotalAbsenceRate, 
           Suspensions, SuspensionSessions, Exclusions,
           KS4_ENTRIES, KS4_PASS_LEV1_PTQ_EE, KS4_PASS_LEV2_PTQ_EE, 
           Label) |>
    rename(Ethnicity = CombinedEthnicity, 
           DOB = birth_date, 
           Gender = gender, 
           ever_FSM = ever_fsm, 
           KS4_Entries = KS4_ENTRIES,
           KS4_Passes_Level1 = KS4_PASS_LEV1_PTQ_EE, 
           KS4_Passes_Level2 = KS4_PASS_LEV2_PTQ_EE,
           Destination = Label)
```

```r
person_absence_cin_excl_ks4_dest_clean
```
